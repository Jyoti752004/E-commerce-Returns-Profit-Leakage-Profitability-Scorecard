```python
import pandas as pd
```

```python
Raw_data = "ecommerce_orders_dataset.csv"
df = pd.read_csv(Raw_data)
print(df.shape)
print(df.head())
```

```python
df.columns = df.columns.str.lower()
```

## Looking for null values in critical columns

```python
df.isnull().sum()
```

```python
critical_col = ['order_id', 'order_date', 'discount_percent', 'discount_amount', 'shipping_cost', 'returned', 'order_amount', 'profit_amount', 'profit_margin_percent']
missing_critical = df[critical_col].isnull().sum()
missing_critical = missing_critical[missing_critical > 0]

if len(missing_critical) > 0:
    print("Missing values found in critical columns:", missing_critical)
else:
    print("No missing values found in critical columns.")
```

## Handling Duplicates

```python
dup_count = df['order_id'].duplicated().sum()
print(dup_count)
if dup_count > 0:
    df = df.drop_duplicates(subset = 'order_id', keep = 'first')
    print("drop duplicates, new shape", df.shape)
else:
    print("No duplicates found in 'order_id' column.")
```

## Standardize categorical fields

```python
categorical_cols = ['country', 'city', 'customer_segment', 'product_category', 'product_subcategory', 'brand', 'payment_method', 'device_type', 'traffic_source', 'membership_status', 'shipping_method', 'warehouse_region', 'order_status', 'season']

for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()
```

## Pharse Date

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors = "coerce")
```

## Creating Derived Columns

1. Boolean Return Flag

```python
df["is_returned"] = df["returned"].str.strip().str.lower().isin(['yes', 'true', '1'])
```

2. Net Revenue

```python
df["net_revenue"] = (df["unit_price"] * df["quantity"] - df["discount_amount"]).where(~df["is_returned"], 0)
print(df.net_revenue.head())
```

3. Discount Band

```python
def discount_band(pct):
    if pct == 0:
        return "0%"
    elif pct <= 10:
        return "1-10%"
    elif pct <= 20:
        return "11-20%"
    elif pct <= 30:
        return "21-30%"
    else:
        return "31%+"
    
df["discount_band"] = df["discount_percent"].apply(discount_band)
print(df.discount_band.head())
```

4. Delivery Band

```python
def delivery_band(days):
    if days <= 3:
        return "0-3 days"
    elif days <= 7:
        return "4-7 days"
    elif days <= 10:
        return "8-10 days"
    else:
        return "10+ days"

df["delivery_band"] = df["delivery_days"].apply(delivery_band)
print(df.delivery_band.head())
```

5. Shipping Cost Ratio

```python
df["shipping_cost_ratio"] = df["shipping_cost"] / df["order_amount"]
print(df.shipping_cost_ratio.head())
```

## Exported Clean File

```python
OUT_PATH = "ecommerce_orders_2026_cleaned.csv"
df.to_csv(OUT_PATH, index = False)
print("Final Shape", df.shape)
```
